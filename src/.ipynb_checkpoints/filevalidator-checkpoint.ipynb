```python
# Example 2: Validate Python module before import
@filecheck(max_size_mb=1, expected_type=FileType.PYTHON, expected_extension='py', validate_malware=True)
def safely_import_module(filepath):
    import importlib.util
    spec = importlib.util.spec_from_file_location("dynamic_module", filepath)
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module
```

```python
# Example 3: Validate text file for processing
@filecheck(max_size_mb=10, expected_type=FileType.TXT, expected_extension='txt')
def process_text_file(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        return f.readlines()
```
